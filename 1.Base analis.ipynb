```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
TW_500= pd.read_csv(
    "../classification/Twitter/Relative_labeling/sigma=500/Twitter-Relative-Sigma-500.data",
    sep=",",
    header=None
)

TW_1000= pd.read_csv(
    "../classification/Twitter/Relative_labeling/sigma=1000/Twitter-Relative-Sigma-1000.data",
    sep=",",
    header=None
)
TW_1500= pd.read_csv(
    "../classification/Twitter/Relative_labeling/sigma=1500/Twitter-Relative-Sigma-1500.data",
    sep=",",
    header=None
)
```

```python
groups = ["NCD", 'AI', 'AS(NA)', 'BL',
         'NAC', 'AS(NAC)', 'CS', 'AT', 'NA','ADL', 'NAD']

columns = []
for group in groups:
    for t in range(7):
        columns.append(f"{group}_{t}")

columns.append("label") 

TW_500.columns = columns
TW_1000.columns = columns
TW_1500.columns = columns

TW_500.head(2)
```

```python
TW_500['label'].value_counts()
TW_1000['label'].value_counts()
TW_1500['label'].value_counts()

balance_df = pd.DataFrame({
    'TW_500': TW_500['label'].value_counts(),
    'TW_1000': TW_1000['label'].value_counts(),
    'TW_1500': TW_1500['label'].value_counts()
}).T

balance_df.plot(kind='bar', figsize=(8,5))

plt.title("Class distribution in Twitter dataset datasets")
plt.xlabel("Dataset")
plt.ylabel("Number of samples")
plt.legend(title="Label")
plt.show()
```

Prevažná väčšina pozorovaní patrí do triedy 0, zatiaľ čo podiel triedy 1 je výrazne nižší ako 3 %.
Vzhľadom na to, že súbory údajov sú nevyvážené, okrem presnosti sa budú používať aj hodnotiace metriky, ako napríklad F1-skóre, presnosť a recall.
